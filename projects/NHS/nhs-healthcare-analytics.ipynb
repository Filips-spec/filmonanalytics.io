```python
#import the libraries
import matplotlib.pyplot as plt
import seaborn as sns

#create a barplot with the total appointments per region
plt.figure(figsize=(14, 10))

#group by region_ons_code
ad_grouped_region = ad.groupby('region_ons_code')['count_of_appointments'].sum().reset_index()

#map region code to to actual region names
region_name_mapping = {'E12000001': 'North East',
                       'E40000011': 'Midlands',
                       'E40000012': 'North East and Yorkshire',
                       'E40000005': 'South East',
                       'E40000003': 'London',
                       'E40000010': 'North West',
                       'E40000007': 'East of England',
                       'E12000006': 'South West'}
#add new column with region names
ad_grouped_region['region_name'] = ad_grouped_region['region_ons_code'].map(region_name_mapping)

#keep only necessary columns and sort
ad_grouped_region = ad_grouped_region[['region_name', 'count_of_appointments']].sort_values(by='count_of_appointments', ascending=False)

# Create the barplot
ax = sns.barplot(
    data=ad_grouped_region,
    x='region_name',
    y='count_of_appointments',
    color='blue')

# Add labels on top of each bar
for p in ax.patches:
    height = p.get_height()
    ax.text(x=p.get_x() + p.get_width() / 2,y=height + 50000, s=f'{int(height):,}',
            ha='center',va='bottom',fontsize=16)

# Plot settings
plt.xticks(rotation=45)
plt.title("Total Appointments by Region", fontsize=18)
plt.xlabel("NHS Region Subdivision", fontsize=15)
plt.ylabel("Total Appointments", fontsize=15)
plt.tight_layout()
plt.savefig("Total appointments by region.jpg", format='jpeg', dpi=300);
plt.show()
```
